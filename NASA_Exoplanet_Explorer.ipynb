```python
## IMPORT RELEVANT APPS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
## IMPORT DATASET
url = (
    "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?"
    "query=select+pl_name,pl_rade,pl_insol,st_teff,"
    "sy_dist+from+pscomppars&format=csv"
)

df = pd.read_csv(url) # create a dataframe from the exoplanet data
print(df.head()) #view first 5 rows of the dataset
print("Total planets:", len(df)) #how many planets do we have?
```

### Column meanings:<br>
---



**pl_name:** planet name<br>

**pl_rade:** planet radius in Earth radii<br>
How big is it, compared to Earth?<br>
Note:
The planet's radius can indicate what kind of planet it might be and therefore whether it might be habitable! <br>
* <0.8 Earth radii -> Their small size means a lower gravitational pull and so they struggle to hold onto an atmosphere.
* 0.8-1.5 Earth radii -> big enough to hold an atmosphere, small enough that they haven't become gas giants
* \>1.5 Earth radii -> usually gas giants.<br>

Remember radius /= mass so this is a bit oversimplified.

**pl_insol:** energy received from its star, relative to Earth.<br>
Note: Insolation can tell us about how much light/heat a planet recieves.
If it's too low, it is unlikely that liquid water would be able to form, making the planet inhabitable to alien life.<br>
The amount of energy received doesn't tell us how warm the planet is definitively, however, as other factors can affect this (e.g. albedo - how reflective the planet is- and atmosphere - how well can it trap energy inside/close to the planet?)

**st_teff:** host star's effective temp in Kelvin<br>
Note: Dimmer stars (e.g. red dwarfs which are <0.8 the mass of the sun) will emit less energy, meaning planets orbiting them will need to orbit more closely to recieve Earth-like amounts of energy. <br>
Our sun's effective temp is: 5,772 K

**sy_dist:** distance from Earth in parsecs<br>

```python
## CLEANING THE DATA WITH PANDAS

columns = [
    "pl_name", "pl_rade", "pl_insol", "st_teff", "sy_dist"
]

# Filter out NULL values:
clean = df[columns].dropna().copy()
# cleaning = in the dataframe columns, remove rows with NULL values.
# Creating a copy so we don't accidentally change values in the original dataframe.

#Filter out impossible values:
clean = clean[
    (clean["pl_rade"]>0) &
    (clean["pl_insol"]>0) &
    (clean["st_teff"]>0) &
    (clean["sy_dist"]>0)
]
# This step filters out impossible values (e.g. negative data points) which would be missed by our first cleaning sweep which looks only for NULL values.

# Review this code later when the website is back up to make sure you haven't made any silly mistakes:
# Check for missing values:
print("Missing values count:")
print(clean.isnull().sum())

# Check minimum values of columns (min should be > 0)
print("Data summary (check min row):")
print(clean[["pl_rade", "pl_insol", "st_teff", "sy_dist"]].describe().loc["min"])

print("Clean observations:", len(clean))
```
