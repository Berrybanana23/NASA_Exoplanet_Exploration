```python
# IMPORT RELEVANT APPS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# IMPORT DATASET
url = (
    "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?"
    "query=select+pl_name,pl_rade,pl_insol,st_teff,"
    "sy_dist+from+pscomppars&format=csv"
)

exoplanet_df = pd.read_csv(url)
print(exoplanet_df.head()) #view first 5 rows of the dataset
print("Total planets:", len(exoplanet_df)) #how many planets do we have?
```

## Column meanings:<br>
**pl_name:** planet name<br>
**pl_rade:** planet radius in Earth radii<br>
**pl_insol:** energy received from its star, relative to Earth<br>
**st_teff:** host star's effective temp in Kelvin<br>
**sy_dist:** distance from Earth in parsecs<br>
